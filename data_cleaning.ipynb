```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
winter = "الشتاء"
spring = "الربيع"
summer = "الصيف"
autumn = "الخريف"
date = "التاريخ"
month = "الشهر"
season = "الفصل"
```

```python
money = pd.read_csv("pos.csv")
```

```python
money.head()
```

```python

money['اسم البند'] = money['اسم البند'].str.strip()
print(money['اسم البند'])
```

making the date index

```python
money.set_index("اسم البند",inplace=True)
money = money.transpose()
print(money.index)



```

```python

money.index = pd.to_datetime(money.index,format="%d/%m/%Y")

money = money.reset_index()
money = money.rename(columns={"index": date})
print(money.head())
```

droping last caolums since all the value in are NaN

```python
empty_col = 'الوحدة :ألف    العملة :ريال'

money[empty_col].isna().sum()
money = money.drop(columns=empty_col)
```

```python

print(money.isna().sum())
```

```python
cols = money.columns.tolist()

seen = False
for i, col in enumerate(cols):
    if col == "أخرى" and not seen:
            cols[i] = "اخرى نقل"
            seen =True
    elif col == "أخرى" and seen:
          cols[i] = "اخرى عام"
          break
        

money.columns = cols
```

the data frame is ready for analysis

```python


money.info()
```

```python
money[month] = money[date].dt.month



sns.countplot(data = money, x= month)
```

```python

mapping= {
    1: winter,
    2: winter,
    3: spring,
    4: spring,
    5: spring,
    6: summer,
    7: summer,
    8: summer,
    9: autumn,
    10: autumn,
    11: autumn,
    12: winter
}
money[season] = money[month].replace(mapping)
```

```python
money.info()
```

```python
sns.countplot(data = money,x=season)
```

```python
print(money[date].min(),money[date].max())
work = "فترة الدراسة"
vic= "اجازة قصيرة"
ram = 'رمضان وعيد الفطر'
summer_vic = "الاجازة الصيفية"
year_split = ["2025-08-23","2025-09-20","2025-09-27","2025-11-20","2025-11-29","2026-01-08","2026-01-17"
              ,"2026-02-17","2026-03-28","2026-05-21","2026-06-01","2026-06-25","2026-08-25"]
year_lable = [work,vic,work,vic,work,vic,work,ram,work,vic,work,summer_vic]
year_split = pd.to_datetime(year_split)
type = "الموسم"
money[type] = pd.cut(money[date],bins=year_split,labels=year_lable,ordered=False)
```

```python
sns.countplot(data=money,x=type)
```

```python

# the dataset have the number with written is 1 thoused unit
# that is 10 is infact 10000
cols =  money.select_dtypes("float").columns
for col in cols:
    print(col)
    money[col] = money[col] *1000
```

```python
(money.select_dtypes('float').sum()).sort_values().plot(kind='bar')
```
