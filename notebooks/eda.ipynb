```python
plt.figure(figsize=(14, 6))
a = pos[key_sectors]
corr = a.corr(numeric_only=True)
sns.heatmap(corr,annot =True,fmt=".1f")
```

```python
a_season = pos[key_sectors].groupby(season)

for name, group in a_season:
    plt.figure()
    plt.figure(figsize=(14, 6))
    plt.title(f"قياس العلاقة في موسم {name} ")
    corr = group.corr(numeric_only=True)
    sns.heatmap(corr,annot =True,fmt=".1f")
```
